# MRL Compute-Accuracy Trade-off

This notebook visualizes the trade-off between **Embedding Size (storage/compute cost)** and **Accuracy (Accuracy/mAP)** when applying Matryoshka Representation Learning (MRL).

```python
import matplotlib.pyplot as plt
import numpy as np

%matplotlib inline
%config InlineBackend.figure_format='retina'
```

## 1. Load data

*(Replace the mock data arrays below with your real benchmark outputs)*

```python
# Vector sizes (MRL Dimensions)
dimensions = [16, 64, 128, 256, 512, 768]

# Corresponding accuracies (e.g., Top-1 Accuracy or mAP)
# NOTE: These are mock values; replace with real outputs from benchmark.py
accuracy = [49.01, 54.70, 55.25, 56.77, 57.27, 56.03]
map_score = [62.11, 67.38, 67.92, 69.18, 69.41, 68.74]
```

## 2. Plot the Trade-off

```python
plt.figure(figsize=(10, 6))

# Plot Accuracy line (blue)
plt.plot(dimensions, accuracy, marker='o', linewidth=2, markersize=8, label='Top-1 Accuracy (%)', color='#1f77b4')

# Plot mAP line (orange, dashed)
plt.plot(dimensions, map_score, marker='s', linewidth=2, markersize=8, label='mAP@5 (%)', color='#ff7f0e', linestyle='--')

# Annotate numeric values on each point for easier comparison
for x, y in zip(dimensions, accuracy):
    plt.annotate(f'{y}', (x, y), textcoords="offset points", xytext=(0,10), ha='center', fontsize=9)
    
for x, y in zip(dimensions, map_score):
    plt.annotate(f'{y}', (x, y), textcoords="offset points", xytext=(0,-15), ha='center', fontsize=9)

# Format plot
plt.title('Compute-Accuracy Trade-off in Matryoshka Representation Learning', fontsize=14, fontweight='bold', pad=20)
plt.xlabel('Representation Size (Number of Dimensions)', fontsize=12)
plt.ylabel('Performance Metric (%)', fontsize=12)

# Set X-axis to log2 scale so small dimensions (16, 64) are not cramped
plt.xscale('log', base=2)
plt.xticks(dimensions, labels=[str(d) for d in dimensions])

plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(loc='lower right', fontsize=11)

# Save high-resolution image and display
plt.tight_layout()
plt.savefig('mrl_tradeoff.png', dpi=300)
plt.show()
```
